# Example use of transcriptionist class

Expects input of an audio file path, and returns a transcription result object in JSON format, described below:



```python
# setup python system path if needed
import sys, json
from pathlib import Path

parent_dir = Path.cwd().resolve().parent
ffmpeg_bin_path = parent_dir / "ffmpeg" / "bin"
sys.path.append(ffmpeg_bin_path)

def print_formatted_sys_path():
    # Rather than deal with the raw output of sys.path, we can use the json module to spit out a nicely formatted object
    formatted_path = json.dumps(sys.path, indent=4)

    # use some colors to make it pretty!
    print("\033[1;34m[SYS.PATH]\033[0m")  # ANSI code for blue for the title
    print("\033[1;32m" + formatted_path + "\033[0m")  # ANSI code for green for paths, then ANSI code for reset


# Call the function to print the formatted sys.path
print_formatted_sys_path()
```

```python
# basic class imports
from AudioTranscriber import AudioTranscriber
from TranscriptTranslator import TranscriptTranslator

source_audio_file = "sample_data/sample_audio.mp4"
output_file = "sample_data/sample_transcription.json"

# init transcriber with input audio file
transcriber = AudioTranscriber(source_audio_file)

# init transcription pipeline
transcriber.transcribe()

# stash JSON obj in memory to hand off to translator
transcription = transcriber.to_json()

# init translator with transcription
translator = TranscriptTranslator(transcription)

# sets input and output languages to english (en)
translator.set_languages("en", "en")

# kick off the translation pipeline process
translator.translate()

# barf translated transcript dict/JSON to variable, won't be needed in actual project (I think...), we'll just pass the obj directly into the next phase of the pipeline
translation = translator.to_json()

# dump the translation dict/JSON to a formatted text file (again, likely won't be needed in actual project)
translator.to_file(output_file)

# let's see the results
print(translation)
```
