```python
import logging

logging.basicConfig()
logging.getLogger("faster_whisper").setLevel(logging.DEBUG)

from faster_whisper import WhisperModel

model_size = "tiny" # "tiny-en" ?

# Run on GPU with FP16
model = WhisperModel(model_size, device="cpu", compute_type="int8") # was originally cuda, will have to put it a env check to determine if cuda is availbalbe

# or run on GPU with INT8
# model = WhisperModel(model_size, device="cuda", compute_type="int8_float16")
# or run on CPU with INT8
# model = WhisperModel(model_size, device="cpu", compute_type="int8")

segments, info = model.transcribe(f"sample_data/Heat_restaurant_scene.mp3", beam_size=5)

print("Detected language '%s' with probability %f" % (info.language, info.language_probability))

for segment in segments:
    print("[%.2fs -> %.2fs] %s" % (segment.start, segment.end, segment.text))
    #print(f"[{segment.start} -> {segment.end}] {segment.text} {segment.}")
```
