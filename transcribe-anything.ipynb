# Python library - trasncribe-anything

```python
# setup stuff
from transcribe_anything import transcribe
from pathlib import Path

from pydantic_settings import BaseSettings, SettingsConfigDict

class NotebookSettings(BaseSettings):
    hf_token: str
    model_config = SettingsConfigDict(env_file='.env', env_file_encoding='utf-8')

settings = NotebookSettings()

#print(f"HF_TOKEN env variable: {settings.hf_token}")
```

```python
# user configuration settings
data_dir = "dummy_data"
audio_file = "sample_audio.mp3"
output_dir = "output"

HF_token = settings.hf_token
```

```python
file_path = Path(data_dir) / audio_file
output_path = Path(output_dir)

result = transcribe(
    url_or_file = str(file_path),
    output_dir = str(output_path),
    model = "tiny", # tiny, small, medium, large
    task = "transcribe", # translate or transcribe
    language = None, # auto detect
    device = "cuda", # cuda, cpu, insane, mlx
    embed = False, # disable subtitle burning
    hugging_face_token = HF_token, # to enable speaker.json diarization
    other_args = [], # additional params for whisper
    initial_prompt = None # no custom prompt needed
)
result
```
